## Exercício 1: O Mistério da Tupla Unitária  
---  
<BR>
<p align='justify'> <b>Contexto:</b> Um estudante de programação decidiu criar uma estrutura para armazenar de forma imutável a porta padrão de um servidor web (apenas o número 80). Ele tentou verificar o tipo de dado da variável usando o comando type() para ter certeza de que havia criado uma tupla e protegido a informação contra alterações acidentais.  Execute e observe a saída do código abaixo: </p>

```
porta_servidor = (80)
print(type(porta_servidor))
```

**1. Ao executar esse script, qual será o tipo de dado impresso no console? Marque a resposta correta.**

A) `<class 'tuple'>`  

B) `<class 'int'>`  (X)

C) `<class 'list'>`  

D) `<class 'syntax_error'>`

<BR>  

**DICA:** Em Python, o que realmente define uma tupla é a presença da vírgula, e não os parênteses (que servem apenas para agrupamento sintático). Para criar uma tupla com um único elemento, é obrigatório colocar uma vírgula após o valor, como em `(80,)`. Sem a vírgula, o Python interpreta a expressão apenas como um número inteiro isolado entre parênteses.

### Exercício 3: Criação Flexível de Tuplas  
---  
<BR>

<p align='justify'><b>Contexto:</b> Um desenvolvedor iniciante estava analisando o código de um sistema legado e se deparou com uma atribuição de dados estruturados que não utilizava parênteses, colchetes ou chaves na declaração. Execute e examine a seguinte linha de código:</p>

```
dados_sistema = "Admin", 1920, 1080
print(type(dados_sistema))
```

**3. Qual tipo de objeto foi armazenado na variável `dados_sistema`? Marque a resposta correta.**  

A) Uma string, pois o primeiro elemento é um texto.  

B) Uma lista (`list`), porque armazena múltiplos tipos de dados de forma sequencial.  

C) Um dicionário (`dict`), devido à associação oculta de dados heterogêneos.  

D) Uma tupla (`tuple`), pois os valores foram separados por vírgulas. (X)

<BR>  

**DICA:** Conforme a sintaxe oficial do Python, a tupla é empacotada automaticamente pelo uso de vírgulas separando elementos. O uso de parênteses na criação de tuplas é opcional na maioria dos cenários (sendo obrigatório só em casos específicos, como na criação de tupla vazia). Portanto, a sintaxe `1, 2` ou `"A", "B"` cria uma tupla heterogênea.

### Exercício 4: Integridade em Catálogos de Produtos  
---  
<BR>

<p align='justify'><b>Contexto:</b> Um sistema de vendas armazena o catálogo de produtos imutáveis em formato de uma lista contendo várias tuplas. Esse modelo híbrido permite ordenar e filtrar a lista ao mesmo tempo em que protege o ID e o nome de cada registro individual contra alterações acidentais. EXECUTE e veja o código implementado:

```
catalogo = [("P01", "Livro"), ("P02", "Caneta")]

# Operação realizada pelo sistema inclui item
catalogo.append(("P03", "Caderno"))
print(len(catalogo)) # Mede quantidade de itens
```

**4. O que acontecerá após a execução dessa rotina? Marque a resposta correta.**  
A) Ocorre um erro de imutabilidade porque a tupla impede o crescimento do catálogo.  

B) A operação é válida e o tamanho final impresso no console será `3`.  (X)

C) O elemento `("P03", "Caderno")` substitui o primeiro registro da lista.  

D) A lista é convertida em uma tupla gigante contendo três pares de strings.  
<BR>

**DICA:** Embora as tuplas do catálogo sejam individualmente rígidas e imutáveis, a estrutura externa que as envolve é uma lista (`[...]`), que possui comportamento dinâmico e flexível. Portanto, podemos adicionar, remover ou reordenar as tuplas nessa lista usando métodos como `.append()`, sem violar a integridade interna de cada tupla individual.
